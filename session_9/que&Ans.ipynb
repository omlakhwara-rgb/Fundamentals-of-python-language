```python
"""
1.Create a Python list called my_fav_apps containing 5 apps you use daily (for example: 'Instagram', 'Zomato', 'Spotify', etc.) and print the list.
"""

l1 = ['Instagram', 'Zomato', 'Spotify','Snapchat','Dominos']

print(l1)
```

```python
"""
2.Use the append() method to add a new app (one you started using recently) to your my_fav_apps list, then print the updated list.
"""

l1 = ['Instagram', 'Zomato', 'Spotify','Snapchat','Dominos']

l1.append("La-pinoz")
print(l1)
```

```python
"""
3.Insert 'WhatsApp' at the second position in your my_fav_apps list using the insert() method and print the result.<br><br><em><strong>Hint:</strong> Remember, list indices start from 0.</em>
"""
l1 = ['Instagram', 'Zomato', 'Spotify','Snapchat','Dominos']

l1.insert (2,"Whatshapp")

print(l1)
```

```python
"""
4.Remove an app you no longer use from your my_fav_apps list using the remove() method, then use the pop() method to remove the last app and print the list after each operation.
"""     

l1 = ['Instagram', 'Zomato', 'Spotify','Snapchat','Dominos']

l1.remove('Zomato')

print(l1)

l1.pop()

print(l1)
```

```python
"""
5.Sort your my_fav_apps list in alphabetical order using the sort() method, then reverse the order to show your least used app first and print both results.<br><br><em><strong>Constraint:</strong> Use only list methods, not built-in sorted().</em>
"""

l1 = ['Instagram', 'Zomato', 'Spotify', 'Snapchat', 'Dominos']

l1.sort()
print(l1)

l1.reverse()
print(l1)
```
